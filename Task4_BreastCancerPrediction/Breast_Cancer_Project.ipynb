```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_breast_cancer

# Load dataset
data = load_breast_cancer()

# Convert to DataFrame (table format)
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target  # 0 = Malignant, 1 = Benign

# View first 5 rows
df.head()
```

```python
# Shape — how many rows and columns?
print("Shape:", df.shape)
```

```python
# Any missing values?
print("Missing values:\n", df.isnull().sum())
```

```python
# Class distribution — how many benign vs malignant?
print("Target count:\n", df['target'].value_counts())
```

```python
# Basic statistics
df.describe()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

#count  plot
sns.countplot(x = 'target', data = df)
plt.title("Benign(1) vs Malignant(0)")
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(14,10))
sns.heatmap(df.corr(), cmap='coolwarm')
plt.title("Feature Correlation")
plt.show()
```

```python
# X = input features (30 columns), y = output label (target)
X = df.drop('target', axis=1)
y = df['target']

print("X shape:", X.shape)   # Should be (569, 30)
print("y shape:", y.shape)   # Should be (569,)
print("\nSample y values:", y.value_counts())
# 1 = Benign, 0 = Malignant
```

```python
from sklearn.model_selection import train_test_split

# 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.2,      # 20% goes to testing
    random_state=42     # fixes randomness so results are reproducible
)

print("Training samples:", X_train.shape[0])   # ~455
print("Testing samples :", X_test.shape[0])    # ~114
print("Training label split:\n", y_train.value_counts())
```

```python
from sklearn.preprocessing import StandardScaler

# Create the scaler
scaler = StandardScaler()

# Fit on training data and transform it
X_train = scaler.fit_transform(X_train)

# Only transform test data (NOT fit again — very important!)
X_test = scaler.transform(X_test)

print("Scaling done ✅")
print("Sample scaled value:", X_train[0][:5])  # Should show small +/- numbers
```

```python
#Quick Verification
import numpy as np

print("=" * 40)
print("✅ DATA PREPROCESSING SUMMARY")
print("=" * 40)
print(f"Total samples       : 569")
print(f"Training samples    : {X_train.shape[0]}")
print(f"Testing samples     : {X_test.shape[0]}")
print(f"Number of features  : {X_train.shape[1]}")
print(f"X_train mean (≈0)   : {np.mean(X_train):.4f}")
print(f"X_train std  (≈1)   : {np.std(X_train):.4f}")
print("=" * 40)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (accuracy_score, 
                              classification_report, 
                              confusion_matrix)

print("All libraries imported ✅")
```

```python
# Define all 4 models in a dictionary
models = {
    "Logistic Regression" : LogisticRegression(max_iter=10000, random_state=42),
    "SVM"                 : SVC(random_state=42),
    "Random Forest"       : RandomForestClassifier(n_estimators=100, random_state=42),
}

# Store results
results = {}

# Loop through each model
for name, model in models.items():
    # 1. Train the model
    model.fit(X_train, y_train)
    
    # 2. Predict on test data
    y_pred = model.predict(X_test)
    
    # 3. Calculate accuracy
    acc = accuracy_score(y_test, y_pred)
    results[name] = acc
    
    # 4. Print results
    print("=" * 50)
    print(f"🤖 Model     : {name}")
    print(f"📊 Accuracy  : {acc * 100:.2f}%")
    print("-" * 50)
    print("Classification Report:")
    print(classification_report(y_test, y_pred, 
          target_names=["Malignant", "Benign"]))
```

```python
print("\n")
print("=" * 40)
print("   📊 MODEL ACCURACY SUMMARY")
print("=" * 40)
for name, acc in results.items():
    bar = "█" * int(acc * 20)   # visual bar
    print(f"{name:<22}: {acc*100:.2f}% {bar}")
print("=" * 40)
best = max(results, key=results.get)
print(f"🏆 Best Model : {best} ({results[best]*100:.2f}%)")
print("=" * 40)
```

```python
import seaborn as sns
from sklearn.metrics import confusion_matrix

# Get SVM predictions
svm_model = models["SVM"]
y_pred_svm = svm_model.predict(X_test)

# Build confusion matrix
cm = confusion_matrix(y_test, y_pred_svm)

# Plot it
plt.figure(figsize=(6, 5))
sns.heatmap(cm, 
            annot=True,          # show numbers inside
            fmt='d',             # integer format
            cmap='Blues',
            xticklabels=['Malignant', 'Benign'],
            yticklabels=['Malignant', 'Benign'],
            linewidths=1,
            linecolor='black')

plt.title("Confusion Matrix — SVM (Best Model)", fontsize=13, fontweight='bold')
plt.ylabel("Actual Label",    fontsize=11)
plt.xlabel("Predicted Label", fontsize=11)
plt.tight_layout()
plt.show()

# Print what the numbers mean
tn, fp, fn, tp = cm.ravel()
print("=" * 45)
print("       🔍 CONFUSION MATRIX BREAKDOWN")
print("=" * 45)
print(f"✅ True Negative  (Correctly said Malignant) : {tn}")
print(f"✅ True Positive  (Correctly said Benign)    : {tp}")
print(f"⚠️  False Positive (Said Benign, was Malignant): {fp}")
print(f"⚠️  False Negative (Said Malignant, was Benign): {fn}")
print("=" * 45)
print(f"🚨 Missed Cancer Cases (FN) : {fn}  ← most critical!")
print("=" * 45)
```

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer

# Get feature names
data = load_breast_cancer()

# Get importance scores from Random Forest
rf_model = models["Random Forest"]
feat_importance = pd.Series(
    rf_model.feature_importances_, 
    index=data.feature_names
).sort_values(ascending=True)

# Plot top 10 features
plt.figure(figsize=(9, 6))
feat_importance.nlargest(10).plot(
    kind='barh', 
    color='teal',
    edgecolor='black'
)
plt.title("Top 10 Most Important Features (Random Forest)", 
          fontsize=13, fontweight='bold')
plt.xlabel("Importance Score")
plt.tight_layout()
plt.show()

# Print top 5
print("\n🔝 Top 5 Most Important Features:")
print("-" * 35)
for i, (feat, score) in enumerate(feat_importance.nlargest(5).items(), 1):
    print(f"{i}. {feat:<30} : {score:.4f}")
```
